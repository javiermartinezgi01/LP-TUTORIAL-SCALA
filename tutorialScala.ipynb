## Revisión previa en Codespaces: Python

Antes de ejecutar el notebook desde un CodeSpace, revisa que la extensión de **Python** esté instalada y actualizada.
Esto ayuda a que la selección de kernel y la ejecución de celdas funcionen correctamente.

```scala
%%bash
set -euo pipefail

echo "Instalando Coursier (cs) si hace falta..."
if ! command -v cs >/dev/null 2>&1; then
  curl -fL "https://github.com/coursier/launchers/raw/master/cs-x86_64-pc-linux.gz" -o /tmp/cs.gz
  gzip -d -f /tmp/cs.gz
  chmod +x /tmp/cs
  sudo mv /tmp/cs /usr/local/bin/cs
fi

cs version
```

```scala
%%bash
set -euo pipefail

echo "Instalando Almond (kernel Scala) con Coursier..."
cs install --contrib almond

ALMOND_BIN="${HOME}/.local/share/coursier/bin/almond"
if [ -x "${ALMOND_BIN}" ]; then
  "${ALMOND_BIN}" --install --force
else
  almond --install --force
fi

echo "Kernel registrado correctamente."
```

```scala
%%bash
set -euo pipefail

echo "Kernels disponibles en Jupyter:"
jupyter kernelspec list
```

## Ejemplo Scala con datos y transformaciones

Este ejemplo usa `case class`, filtrado, agrupación y cálculos sobre una colección pequeña.

```scala
case class Compra(usuario: String, categoria: String, monto: Double)

val compras = Seq(
  Compra("Ana", "libros", 24.5),
  Compra("Luis", "software", 58.0),
  Compra("Ana", "software", 19.9),
  Compra("Marta", "libros", 31.0),
  Compra("Luis", "cursos", 42.5),
  Compra("Ana", "cursos", 15.0)
)

val comprasCaras = compras.filter(_.monto >= 20)
val totalPorCategoria = compras.groupBy(_.categoria).view.mapValues(_.map(_.monto).sum).toMap
val totalPorUsuario = compras.groupBy(_.usuario).view.mapValues(_.map(_.monto).sum).toMap

println(s"Compras con monto >= 20: ${comprasCaras.size}")
println(s"Totales por categoría: $totalPorCategoria")
println(s"Totales por usuario: $totalPorUsuario")
println(s"Usuario con mayor gasto: ${totalPorUsuario.maxBy(_._2)}")
```
