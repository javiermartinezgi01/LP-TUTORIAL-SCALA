## Ejemplo Scala con datos y transformaciones

Este ejemplo usa `case class`, filtrado, agrupación y cálculos sobre una colección pequeña.

```scala
case class Compra(usuario: String, categoria: String, monto: Double)

val compras = Seq(
  Compra("Ana", "libros", 24.5),
  Compra("Luis", "software", 58.0),
  Compra("Ana", "software", 19.9),
  Compra("Marta", "libros", 31.0),
  Compra("Luis", "cursos", 42.5),
  Compra("Ana", "cursos", 15.0)
)

val comprasCaras = compras.filter(_.monto >= 20)
val totalPorCategoria = compras.groupBy(_.categoria).view.mapValues(_.map(_.monto).sum).toMap
val totalPorUsuario = compras.groupBy(_.usuario).view.mapValues(_.map(_.monto).sum).toMap

println(s"Compras con monto >= 20: ${comprasCaras.size}")
println(s"Totales por categoría: $totalPorCategoria")
println(s"Totales por usuario: $totalPorUsuario")
println(s"Usuario con mayor gasto: ${totalPorUsuario.maxBy(_._2)}")
```

# 1. Introducción y primeros pasos

En esta primera sección se presentan las bases del lenguaje: cómo ejecutar un saludo, cuándo usar `val` o `var`, cómo inferir tipos y cómo trabajar con colecciones inmutables comunes.

## 1.1 Hello World

El primer paso en Scala es confirmar que el kernel responde correctamente. Un `println` simple sirve para validar el entorno y la ejecución de Scala en el notebook.

```scala
println("Hello World desde Scala")
```

## 1.2 `val` vs `var`

`val` crea una referencia inmutable y es la opción recomendada por defecto. `var` permite reasignar el valor, pero conviene evitarlo salvo que la mutabilidad sea realmente necesaria.

```scala
val nombre = "Scala"
var contador = 1
contador = contador + 1
println(s"$nombre, contador = $contador")
```

## 1.3 Tipos básicos e inferencia de tipos

Scala puede deducir el tipo de una expresión a partir del valor asignado. Aun así, el lenguaje sigue siendo fuertemente tipado: el compilador conoce el tipo real en tiempo de compilación.

```scala
val entero = 42
val decimal = 3.14
val activo = true
val saludo = "Hola"
println(s"Tipos inferidos: entero=$entero, decimal=$decimal, activo=$activo, saludo=$saludo")
```

## 1.4 Colecciones básicas: `List`, `Vector`, `Map`

Las colecciones inmutables son la base del estilo funcional en Scala. `List` es útil para recorridos secuenciales, `Vector` ofrece acceso eficiente por índice y `Map` permite asociar claves con valores.

```scala
val numeros = List(1, 2, 3)
val letras = Vector("a", "b", "c")
val edades = Map("Ana" -> 29, "Luis" -> 34)
println(s"List: $numeros")
println(s"Vector: $letras")
println(s"Map: $edades")
```

# 2. Tour rápido del lenguaje

Esta sección muestra una idea central de Scala: muchas construcciones son expresiones que devuelven un valor. También se introducen `if`, `def`, `case class` y `pattern matching`.

## 2.1 Expresiones: todo retorna un valor en Scala

En Scala, un bloque, un `if` o un `match` producen un resultado. Esto favorece estilos de programación más expresivos y composables que en lenguajes con sentencias puramente imperativas.

```scala
val resultado = if (10 > 5) "mayor" else "menor"
val bloque = {
  val base = 10
  base * 2
}
println(s"if => $resultado, bloque => $bloque")
```

## 2.2 Condicionales: `if/else` como expresión

Un `if` en Scala no solo controla flujo: también devuelve un valor. Eso permite asignar el resultado directamente a una variable sin necesidad de separar la condición del valor obtenido.

```scala
val edad = 17
val etapa = if (edad >= 18) "adulto" else "menor"
println(s"Con $edad años, la etapa es $etapa")
```

## 2.3 Definición de funciones con `def` y diferencia con val-función

`def` declara una función evaluada cuando se invoca. En cambio, un `val` puede almacenar una función como valor, lo que resulta útil cuando la lógica se pasa como argumento o se reutiliza como dato.

```scala
def duplicar(x: Int): Int = x * 2
val triplicar: Int => Int = x => x * 3
println(s"duplicar(4) = ${duplicar(4)}")
println(s"triplicar(4) = ${triplicar(4)}")
```

## 2.4 `Case classes`: qué son y para qué sirven

Las `case class` modelan datos inmutables con poco código. Scala genera automáticamente constructores, acceso a campos, `equals`, `hashCode` y soporte para `pattern matching`.

```scala
case class Persona(nombre: String, edad: Int)
val persona = Persona("Ana", 30)
println(persona)
println(s"Nombre: ${persona.nombre}, edad: ${persona.edad}")
```

## 2.5 Pattern matching: introducción conceptual

`match` permite decidir qué hacer según el valor de una expresión. Es una alternativa más clara a múltiples `if/else` cuando hay varios casos posibles.

```scala
val numero = 2
val descripcion = numero match {
  case 1 => "uno"
  case 2 => "dos"
  case _ => "otro"
}
println(descripcion)
```

# 3. Particularidades del lenguaje

Scala introduce varias herramientas útiles para escribir código expresivo y seguro: inferencia de tipos, evaluación perezosa, `Option` y recursión de cola.

## 3.1 Inferencia de tipos en detalle

La inferencia permite escribir menos código sin perder seguridad. El compilador sigue verificando los tipos y detecta inconsistencias antes de ejecutar el programa.

```scala
val mensaje = "Scala infiere este tipo automáticamente"
val longitud = mensaje.length
println(s"$mensaje tiene $longitud caracteres")
```

## 3.2 Evaluación perezosa: `lazy val`

Un `lazy val` no se calcula al declararse, sino en el primer acceso. Es útil cuando el cálculo es costoso o cuando solo se necesita si realmente se usa.

```scala
lazy val cargaPesada = {
  println("Calculando...")
  99
}
println("Antes del primer acceso")
println(cargaPesada)
println(cargaPesada)
```

## 3.3 `Option`: manejo de ausencia de datos

`Option` evita usar `null` como mecanismo implícito de ausencia. `Some(valor)` representa presencia y `None` representa ausencia de forma explícita.

```scala
def buscarEdad(nombre: String): Option[Int] = if (nombre == "Ana") Some(30) else None
val edadAna = buscarEdad("Ana")
val edadPedro = buscarEdad("Pedro")
println(edadAna.getOrElse(0))
println(edadPedro.getOrElse(0))
```

## 3.4 `Tail recursion` con `@tailrec`

La recursión de cola permite que el compilador optimice llamadas recursivas para evitar crecer la pila de ejecución. La anotación `@tailrec` ayuda a validar que la función realmente cumple esa condición.

```scala
import scala.annotation.tailrec

@tailrec
def suma(n: Int, acumulado: Int = 0): Int =
  if (n == 0) acumulado else suma(n - 1, acumulado + n)

println(suma(5))
```

# 4. Programación funcional con Scala

Aquí se ve el cambio de paradigma más importante: dejar de modificar estado y empezar a transformar datos con funciones puras e inmutabilidad.

## 4.1 De modificar datos a transformarlos

En lugar de cambiar una estructura existente, la programación funcional crea nuevos resultados a partir de datos de entrada. Eso simplifica el razonamiento y reduce efectos secundarios.

```scala
val original = List(1, 2, 3)
val transformada = original.map(_ * 10)
println(s"original = $original")
println(s"transformada = $transformada")
```

## 4.2 Imperativo vs funcional: ejemplo con contador

El estilo imperativo cambia una variable paso a paso. El estilo funcional produce un nuevo valor a partir del anterior, sin alterar el original.

```scala
var contadorImperativo = 0
contadorImperativo = contadorImperativo + 1
val contadorFuncional = 0 + 1
println(s"imperativo = $contadorImperativo, funcional = $contadorFuncional")
```

## 4.3 Inmutabilidad

La inmutabilidad hace que un valor no cambie después de crearse. En Scala, esto es la norma recomendada porque facilita pruebas, concurrencia y mantenimiento.

```scala
val listaBase = List(1, 2, 3)
val listaNueva = listaBase :+ 4
println(s"base = $listaBase")
println(s"nueva = $listaNueva")
```

## 4.4 Funciones puras

Una función pura devuelve siempre el mismo resultado para la misma entrada y no produce efectos secundarios observables.

```scala
def cuadrado(x: Int): Int = x * x
println(cuadrado(5))
println(cuadrado(5))
```

## 4.5 Funciones como valores

Scala permite guardar funciones en variables, pasarlas como argumentos y devolverlas desde otras funciones. Eso abre la puerta a patrones muy flexibles.

```scala
val incrementar: Int => Int = _ + 1
println(incrementar(7))
```

## 4.6 Funciones anónimas

Las funciones anónimas son útiles cuando la lógica es breve y no merece una definición con nombre propio.

```scala
val duplicados = List(1, 2, 3).map(x => x * 2)
println(duplicados)
```

## 4.7 Funciones de orden superior

Una función de orden superior recibe una función como argumento o devuelve una función. Es uno de los pilares del estilo funcional en Scala.

```scala
def aplicarDosVeces(f: Int => Int, valor: Int): Int = f(f(valor))
println(aplicarDosVeces(_ + 1, 10))
```

## 4.8 `map`, `filter` y `reduce`

Estas tres operaciones resumen gran parte del trabajo funcional con colecciones: transformar, filtrar y consolidar resultados.

```scala
val base = List(1, 2, 3, 4)
val dobles = base.map(_ * 2)
val pares = base.filter(_ % 2 == 0)
val suma = base.reduce(_ + _)
println(s"dobles = $dobles, pares = $pares, suma = $suma")
```

## 4.9 `flatMap`

`flatMap` combina transformación y aplanado. Se usa mucho cuando una operación produce colecciones anidadas y se quiere un único nivel de resultados.

```scala
val palabras = List("hola mundo", "scala es elegante")
val tokens = palabras.flatMap(_.split(" ").toList)
println(tokens)
```

# 5. Ejemplo práctico — análisis de notas de estudiantes

Ahora usaremos listas, `case class`, `map`, `filter`, `maxBy` y `pattern matching` en un caso real.

Trabajaremos con una lista de estudiantes y sus calificaciones. El objetivo es responder cinco preguntas clave: identificar aprobados, obtener nombres, calcular promedio, encontrar la mejor nota y clasificar el desempeño.

## 5.1 Presentación del caso

### Modelar datos con `case class`

Una `case class` representa datos de forma concisa, inmutable y expresiva. Scala genera automáticamente el constructor con parámetros nombrados, `toString` legible, igualdad por valor y soporte natural para `pattern matching`.

```scala
case class Estudiante(
  nombre: String, 
  nota: Double
  )

val estudiantes = List(
  Estudiante("Ana", 4.5),
  Estudiante("Luis", 2.8),
  Estudiante("Marta", 3.9),
  Estudiante("Juan", 5.0),
  Estudiante("Sofía", 2.5)
)

println(estudiantes)
```

## 5.2 Filtrar estudiantes aprobados con `filter`

`filter` recorre cada elemento de la lista y conserva únicamente aquellos que cumplen la condición dada. La lista original no se modifica; se crea una nueva lista con los aprobados.

```scala
val aprobados = estudiantes.filter(_.nota >= 3.0)
println(aprobados)
```

## 5.3 Transformar datos con `map`

`map` transforma cada elemento aplicando una función. Aquí extraemos el nombre de cada estudiante aprobado para obtener una nueva lista de valores más compacta.

```scala
val nombresAprobados = aprobados.map(_.nombre)
println(nombresAprobados)
```

## 5.5 Promedio y mejor nota: `sum`, `length`, `maxBy`

Con operaciones sencillas sobre colecciones se calcula el promedio del grupo y se identifica al estudiante con la nota más alta.

```scala
val promedio = estudiantes.map(_.nota).sum / estudiantes.length
val mejor = estudiantes.maxBy(_.nota)
println(f"Promedio = $promedio%.2f")
println(s"Mejor nota = $mejor")
```

## 5.6 Clasificación con `pattern matching`

Aquí se aplica el concepto visto en la sección 2.5. `match` evalúa condiciones de forma legible y exhaustiva, eliminando cadenas largas de `if-else`. Cada caso (`case`) es una rama clara.

```scala
def clasificar(nota: Double): String = nota match {
  case n if n >= 4.5 => "Excelente"
  case n if n >= 3.0 => "Aprobado"
  case _ => "Reprobado"
}

val reporte = estudiantes.map(e => (e.nombre, clasificar(e.nota)))
println(reporte)
```

## 5.7 Aplicar `Option` al ejemplo

Si la lista está vacía, conviene devolver `None` en lugar de fallar con una excepción. Así la siguiente operación decide explícitamente qué hacer ante la ausencia de datos.

```scala
def mejorEstudiante(lista: List[Estudiante]): Option[Estudiante] =
  lista.sortBy(_.nota).lastOption

println(mejorEstudiante(estudiantes))
println(mejorEstudiante(Nil))
```

## 5.8 Mini reto interactivo: `filter + map` encadenados

Pregunta: ¿qué contiene `resultado` si filtramos a quienes superan la nota mínima y luego extraemos solo sus nombres?

Pausa: piensa la respuesta antes de pasar a la siguiente línea.

```scala
val resultado = estudiantes
  .filter(_.nota >= 3.0)
  .map(_.nombre)

println(resultado)
```
